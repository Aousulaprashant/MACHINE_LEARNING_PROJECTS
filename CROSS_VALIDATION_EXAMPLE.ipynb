```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score

from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
#importing datasets

loan_data = pd.read_csv('/content/loan_dataset.csv')
loan_data.head()
```

```python
from sklearn.preprocessing import LabelEncoder
labelencoder = LabelEncoder()

loan_data['Loan_Status'] = labelencoder.fit_transform(loan_data['Loan_Status'])
loan_data.head()

# 1-> Y
# 0 -> N
```

```python
loan_data.describe()
loan_data.isnull().sum()
```

```python
#droping missing values

loan_data=loan_data.dropna()
```

```python
loan_data.isnull().sum()
```

```python
loan_data['Property_Area'].value_counts()
```

```python
loan_data['Property_Area']=labelencoder.fit_transform(loan_data['Property_Area'])
loan_data['Self_Employed']=labelencoder.fit_transform(loan_data['Self_Employed'])
loan_data['Education']=labelencoder.fit_transform(loan_data['Education'])
loan_data['Married']=labelencoder.fit_transform(loan_data['Married'])
loan_data['Gender']=labelencoder.fit_transform(loan_data['Gender'])
loan_data.head()
```

```python


loan_data=loan_data.replace(to_replace='3+',value=4)
loan_data['Dependents'].value_counts()
```

saparating the DATA

```python


X = loan_data.drop(columns=['Loan_ID', 'Loan_Status'], inplace=False)
Y=loan_data['Loan_Status']
```

```python
print(X.shape)
print(Y.shape)
print(X.dtypes)
```

```python
#models array

models =[LogisticRegression(max_iter=1000),SVC(kernel='linear'),KNeighborsClassifier(),RandomForestClassifier()]


```

```python
print(models)
print(X)
['']
```

```python
cv_score_lr = cross_val_score(LogisticRegression(max_iter=1000), X, Y, cv=5)
print(cv_score_lr)
mean_accuracy_lr = sum(cv_score_lr)/len(cv_score_lr)
mean_accuracy_lr = mean_accuracy_lr*100
mean_accuracy_lr = round(mean_accuracy_lr, 2)
print(mean_accuracy_lr)
```

```python
def compare_models_cross_validation():

  for model in models:

    cv_score = cross_val_score(model, X,Y, cv=5)

    mean_accuracy = sum(cv_score)/len(cv_score)

    mean_accuracy = mean_accuracy*100

    mean_accuracy = round(mean_accuracy, 2)

    print('Cross Validation accuracies for ', model, '=  ', cv_score)
    print('Accuracy % of the ', model, mean_accuracy)
    print('----------------------------------------------')
```

```python
compare_models_cross_validation()
```
