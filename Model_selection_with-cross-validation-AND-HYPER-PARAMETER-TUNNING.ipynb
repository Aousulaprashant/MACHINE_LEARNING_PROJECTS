```python
import pandas as pd
import numpy as np
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
heart_data=pd.read_csv('/content/heart_disease_data.csv')
heart_data.head()
```

# MODEL SELECTION

```python
X=heart_data.drop(columns='target',axis=1)
Y=heart_data['target']
```

creating the list of models

```python
models_list=[LogisticRegression(max_iter=1000),SVC(kernel='linear'),KNeighborsClassifier(),RandomForestClassifier(random_state=0)]
```

- Comparing the models with default hyperparameter values using Cross Validation
- model selection based on cross validation

```python
def compare_models_cross_val():

  for model in models_list:
    cross_scores=cross_val_score(model,X,Y,cv=5)
    avg_cross_score=sum(cross_scores)/len(cross_scores)
    avg_cross_score=avg_cross_score*100
    avg_cross_score=round(avg_cross_score,2)

    print('for',model,'model the accuracy is:',avg_cross_score,'%')
    print('---------------------------------------------------------------')
```

```python
compare_models_cross_val()
```

Comparing the models with different Hyperparameter values using **GridSearchCV**

```python
models_list=[LogisticRegression(max_iter=10000),SVC(),KNeighborsClassifier(),RandomForestClassifier(random_state=0)]
```

```python
#parameters Hyperparameters list
Hyparameters_Lists={
    'LogisticRegression':{
        'C':[1,5,10,20]
    },
    'SVC_parameters':{
        'C':[1,5,10,20],
        'kernel' : ['linear','poly','rbf','sigmoid'],
    },
    'KNN_hyperparameters' : {

        'n_neighbors' : [3,5,10]
    },


    'random_forest_hyperparameters' : {

        'n_estimators' : [10, 20, 50, 100]
    }
}
```

```python

model_keys = list(Hyparameters_Lists.keys())
print(Hyparameters_Lists['LogisticRegression'])
```

```python
def campare_cross_Hyper_parameters(List_of_Models,List_of_hyper_parameters):
  i=0
  result=[]
  for model in List_of_Models:

    key=model_keys[i]
    params=List_of_hyper_parameters[key]

    print(model)
    print(params)
    i+=1
    classifer=GridSearchCV(model,params,cv=5)
    classifer.fit(X,Y)

    result.append({
        'model_name':model,
        'parames':classifer.best_params_,
        'highst_score':classifer.best_score_
    })


  Results=pd.DataFrame(result, columns = ['model_name','parames','highst_score'])
  return Results
```

```python
campare_cross_Hyper_parameters(models_list,Hyparameters_Lists)
```
